I'm writing this notebook to see what's available to access with the new NOAA-CEFI Earthmover organization.

I am a member of the NOAA-CEFI organization on Earthmover. I have conda 26.9.4 installed, I'm using python 3.14.10, and arraylake version 1.8.0. I've already authenticated with arraylake in terminal by running `arraylake auth login`.

```python
from arraylake import Client
import xarray as xr
import zarr
```

Let's assume I didn't know the repo name, I only had the organization. Here's how I could see the repositories:

```python
client = Client()
client.list_repos("noaa-cefi")
```

Now let's assume I know the repository I want is `cefi-nep-hindcast-monthly`. Here's how I could see the datasets:

```python
repo = client.get_repo("noaa-cefi/cefi-nep-hindcast-monthly")
session = repo.readonly_session(branch="main")
root = zarr.open_group(session.store, mode="r", zarr_format=3)
root.tree()
```

Now let's assume I know the group I want is `raw`, and the dataset I want is `main`. Here's how I can see more about the dataset dimensions, coordinates, and variables:

```python
main = xr.open_zarr(session.store, zarr_format=3, group="raw/main")
main
```

Now let's assume I've identified the variable I want is `thetao`. Here's how to see more about that variable:

```python
ds_temp = main[["thetao"]]
ds_temp
```
